### navigation api example

```python
# 0. 导入控制类
from x3_nav_controller import YahboomX3Controller
import rclpy

rclpy.init()
controller = YahboomX3Controller()
# note: must shutdown after using
# rclpy.shutdown()
```

```python
# 1. 获取当前姿态
# 高频里程计姿态（推荐实时监控用）
odom_pose = controller.get_current_pose(use_amcl=False)
print(f"里程计位置：x={odom_pose['x']}, y={odom_pose['y']}, 朝向={odom_pose['yaw']}")

# 高精度AMCL姿态（推荐导航定位用）
amcl_pose = controller.get_current_pose(use_amcl=True)
print(f"AMCL位置：x={amcl_pose['x']}, y={amcl_pose['y']}, 朝向={amcl_pose['yaw']}")
```

```python
# 2. 发送导航目标
# 动作方式（推荐，有反馈）
controller.send_goal_by_action(x=0.0, y=0.0, yaw=0.0)

# 话题方式（备用）
# controller.send_goal_by_topic(x=1.5, y=0.8, yaw=0.0)
```

```python
controller.send_goal_by_action(x=-0.2, y=0.0, yaw=1.57)
```

```python
controller.send_goal_by_action(x=0.5, y=0.0, yaw=1.57)
```

```python
controller.send_goal_by_action(x=0.2, y=0.0, yaw=0.0)
```

```python
# 3. 取消导航
controller.cancel_navigation()
```
